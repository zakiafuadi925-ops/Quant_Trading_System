```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
```

```python
# Fetch data dari yfinance
cryptos  = ['BTC-USD', 'ETH-USD', 'BNB-USD', 'ADA-USD', 'SOL-USD']
print(" Fetching live crypto data...")
crypto_data = {}

for symbol in cryptos:
    print(f"Fetching {symbol}...")
    ticker = yf.Ticker(symbol)
    data = ticker.history(period='30d')
    crypto_data[symbol] = data['Close']

print(f" Succesfully fetched data for {len(cryptos)} cryptocurrencies")
print(f"Date range: {crypto_data['BTC-USD'].index[0].date()} to {crypto_data['BTC-USD'].index[-1].date()}")

# Create combined dataframe
df = pd.DataFrame(crypto_data)
print(f"\nData shape: {df.shape}")
print(f"Latest price:")
for symbol in cryptos:
    latest_price = df[symbol].iloc[-1]
    print(f"{symbol}: ${latest_price:,.2f}")
```

```python
print("Data quality check:")
print(f"Missing Values;")
print(df.isnull().sum())
print(f"\nData Statistics:")
print(df.describe().round(2))

# Show data sample
print(f"\n Recent 5 days:")
print(df.tail())
```

```python
df_normalized = (df / df.iloc[0] - 1) * 100

plt.figure(figsize=(14, 8))
colors = ['#f7931a', '#627eea', '#f3ba2f', '#0033ad', '#9945ff']

for i, symbol in enumerate(cryptos):
    plt.plot(df_normalized.index, df_normalized[symbol],
             label=symbol, linewidth=2.5, color=colors[i])

plt.title('30_Day Crypto Performance Comparison (%)', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price Change(%)', fontsize=10)
plt.legend(loc='best', fontsize=10)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)

# Add latest performance text
latest_change = df_normalized.iloc[-1].round(1)
textstr = 'Latest 30-Days Performance:\n'+'\n'.join([f'{k}: {v:+.1f}%' for k, v in latest_change.items()])
plt.text(0.02, 0.98, textstr, transform=plt.gca().transAxes, fontsize=9,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

plt.tight_layout()
plt.show()

print(" Performance Summary (30 days):")
for symbol, change in latest_change.items():
    print(f"{symbol}: {change:+.1f}%")
```
